Q.1 Equilibrium conversion checker
In a liquid phase reactor the reaction A + B =(Reversible) C reaches equilibrium. Initial concentrations are [A] = 1.0, [B] =1.0 and [C] = 0mol/L of C is formed, Kc = x/(1-x)^2 , which rearranges to the Quadratic equation
Kc.x^2 - (2Kc+1).x +Kc =0

```python
import math


initial_A = 1.0
initial_B = 1.0

try:
    Kc = float(input("Enter the equilibrium constant Kc: "))
except ValueError:
    raise ValueError("Only numerical values should be given.") from None

if Kc <= 0:
    raise ValueError("Kc must be greater than zero.")

# Coefficients for a*x^2 + b*x + c = 0
a = Kc
b = -(2 * Kc + 1)
c = Kc
discriminant = b**2 - 4 * a * c

root_1 = (-b + math.sqrt(discriminant)) / (2 * a)
root_2 = (-b - math.sqrt(discriminant)) / (2 * a)
roots = (root_1, root_2)

physical_roots = [root for root in roots if 0 <= root <= 1]
if not physical_roots:
    raise ValueError("No physically meaningful root was found.")
x = min(physical_roots)

A_equilibrium = initial_A - x
B_equilibrium = initial_B - x
C_equilibrium = x
conversion_percent = x / initial_A * 100

if conversion_percent >= 70:
    classification = "High"
elif conversion_percent >= 40:
    classification = "Moderate"
else:
    classification = "Low"

print(f"Kc = {Kc:g}")
print(f"a = {a:g}, b = {b:g}, c = {c:g}")
print(f"Discriminant = {discriminant:g}")
print(f"Root 1 = {root_1:g}")
print(f"Root 2 = {root_2:g}")
print(f"Physical root (conversion, x) = {x:.3f}")
print(f"[A] at equilibrium = {A_equilibrium:.3f} mol/L")
print(f"[B] at equilibrium = {B_equilibrium:.3f} mol/L")
print(f"[C] at equilibrium = {C_equilibrium:.3f} mol/L")
print(f"Limiting-reactant conversion = {conversion_percent:.3f}%")
print(f"Conversion classification = {classification}")
```

Q.2 Batch reactor simulation and batch tracking
Part A. A batch reactor decomposes reactant A (A -> products) by a first order reaction, dC/dt = -kC. Use the Euler method: C_new = C_old - k * C_old * delta_t. Take C0 = 2.0 mol/L, k = 0.08 min^-1, and delta_t = 1 min.
Use a while loop to print a table of time in minutes, concentration in mol/L, and conversion in percent. Continue until conversion reaches at least 60% or time reaches 100 min. Print the time needed and final concentration. Use math.exp() to calculate the exact concentration at 10 min and print the absolute error of the Euler result at 10 min.

```python
import math

C0 = 2.0
k = 0.08
delta_t = 1.0
target_conversion = 60.0
max_time = 100.0

concentration = C0
time = 0.0
conversion = 0.0
euler_concentration_at_10 = None

print(f"{'Time (min)':>10} {'Concentration (mol/L)':>24} {'Conversion (%)':>18}")
print(f"{time:10.0f} {concentration:24.3f} {conversion:18.2f}")

while conversion < target_conversion and time < max_time:
    concentration = concentration - k * concentration * delta_t
    time += delta_t
    conversion = (C0 - concentration) / C0 * 100
    print(f"{time:10.0f} {concentration:24.3f} {conversion:18.2f}")

    if math.isclose(time, 10.0):
        euler_concentration_at_10 = concentration

exact_concentration_at_10 = C0 * math.exp(-k * 10)
if euler_concentration_at_10 is not None:
    error_at_10 = abs(euler_concentration_at_10 - exact_concentration_at_10)
    print(f"\nExact concentration at 10 min: {exact_concentration_at_10:.6f} mol/L")
    print(f"Euler concentration at 10 min: {euler_concentration_at_10:.6f} mol/L")
    print(f"Absolute error at 10 min: {error_at_10:.6f} mol/L")

if conversion >= target_conversion:
    print(f"\nTarget conversion of {target_conversion:.0f}% reached.")
else:
    print(f"\nMaximum simulation time of {max_time:.0f} min reached.")

print(f"Time needed: {time:.0f} min")
print(f"Final concentration: {concentration:.6f} mol/L")
print(f"Final conversion: {conversion:.2f}%")
```

Q.2 Batch reactor simulation and batch tracking
Part B: Batch labels such as PLANT20261030 are stored as text. writea program that reads a label, loops through each character, and print the number of digits and the sum of the digits. show a message for empty label and for a label with no digits use .isdigit() . Test with PLANT20261030, and empty label and ABCDEF.

```python
def analyze_batch_label(label):
    if label == "":
        print("The batch label is empty.")
        return

    digit_count = 0
    digit_sum = 0

    for character in label:
        if character.isdigit():
            digit = int(character)
            digit_count += 1
            digit_sum += digit

    if digit_count == 0:
        print("No digits were found in the batch label.")
    else:
        print(f"Number of digits: {digit_count}")
        print(f"Sum of digits: {digit_sum}")


test_labels = ["PLANT20261030", "", "ABCDEF"]
for test_label in test_labels:
    print(f"\nTesting label: {test_label!r}")
    analyze_batch_label(test_label)

custom_label = input("\nEnter a batch label: ")
print("Custom label result:")
analyze_batch_label(custom_label)
```

Q.3 Additive screening and sampling schedule
Part A: A laboratory has n candidate additives and wants to test r of them together in one formulation (order does not matter). write a recursive function factorial(n) (remember the base case for 0!) and a function of combination (n,r) = n!/(r!(n-r)!) that uses it. Reject invalid input (negative values, r>n). print the number of possible formulations for n = 6, r = 3, and show the message for n = 5 and r = 7.  

```python
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative values.")
    if n == 0:
        return 1
    return n * factorial(n - 1)


def combination(n, r):
    if n < 0 or r < 0:
        raise ValueError("n and r must be non-negative.")
    if r > n:
        raise ValueError("r cannot be greater than n.")

    return factorial(n) // (factorial(r) * factorial(n - r))


n = 6
r = 3
print(f"Number of formulations for n={n}, r={r}: {combination(n, r)}")

try:
    print(f"Number of formulations for n=5, r=7: {combination(5, 7)}")
except ValueError as error:
    print(f"Invalid input for n=5, r=7: {error}")
```

Part B: During a reaction, samples are taken at times = Fibonacci number (times) 2 min. write an iterative function fib_iter(n) that returns a list of the first n fibonacci numbers, and a recursive function fib_rec(k) that returns the k^th Fibonacci number (F0 = 0, F1 = 1). for n = 10, check that both functions agree and print a table sample No.| Fibonacci tag | Sampling time (min).

```python
def fib_iter(n):
    if n < 0:
        raise ValueError("n must be non-negative.")

    sequence = []
    previous, current = 0, 1
    for _ in range(n):
        sequence.append(previous)
        previous, current = current, previous + current
    return sequence


def fib_rec(k):
    if k < 0:
        raise ValueError("k must be non-negative.")
    if k == 0:
        return 0
    if k == 1:
        return 1
    return fib_rec(k - 1) + fib_rec(k - 2)


n = int(input("Enter the number of Fibonacci values to generate: "))
fibonacci_values = fib_iter(n)
recursive_values = [fib_rec(index) for index in range(n)]

if fibonacci_values == recursive_values:
    print("The iterative and recursive functions agree.")
else:
    raise AssertionError("The iterative and recursive results do not match.")

print(f"{'Sample No.':>10} {'Fibonacci tag':>18} {'Sampling time (min)':>22}")
for sample_number, fibonacci_value in enumerate(fibonacci_values, start=1):
    fibonacci_index = sample_number - 1
    sampling_time = fibonacci_value * 2
    print(f"{sample_number:10} {f'F{fibonacci_index} = {fibonacci_value}':>18} {sampling_time:22}")
```
