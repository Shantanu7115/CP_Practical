```python
# student grade A - above 95 marks, B - above 85 marks, C - above 75 marks, D - above 65 marks, E - above 55 marks, F - 55 or below
# if elif else format
marks = int(input("Enter your marks: ") )
if marks > 95:
    print("Your grade is A")
elif marks > 85:
    print("Your grade is B")
elif marks > 75:
    print("Your grade is C")
elif marks > 65:
    print("Your grade is D")
elif marks > 55:
    print("Your grade is E")
else:
    print("Your grade is F")    
```
